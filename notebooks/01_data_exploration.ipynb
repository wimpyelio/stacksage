# StackSage — Data Exploration
Explore the cleaned QA dataset before indexing.

```python
import json, pandas as pd, matplotlib.pyplot as plt
from pathlib import Path

CLEAN = Path('../data/processed/qa_pairs_clean.jsonl')
records = [json.loads(l) for l in CLEAN.open() if l.strip()]
df = pd.DataFrame(records)
print(f'Records: {len(df)}')
df.head(2)
```

## Score Distribution

```python
fig,axes = plt.subplots(1,2,figsize=(12,4))
df['question_score'].clip(upper=200).hist(bins=40,ax=axes[0]); axes[0].set_title('Question Score')
df['answer_score'].clip(upper=100).hist(bins=40,ax=axes[1]); axes[1].set_title('Answer Score')
plt.tight_layout(); plt.show()
```

## Tag Distribution

```python
from collections import Counter
tag_counts = Counter(t for tags in df['tags_list'] for t in tags)
pd.Series(tag_counts).sort_values(ascending=False).head(20).plot(kind='bar',figsize=(14,4),title='Top 20 Tags')
plt.tight_layout(); plt.show()
```

## Text Length Distribution

```python
df['ans_len'] = df['answer_prose'].str.split().str.len()
df['ans_len'].clip(upper=600).hist(bins=40,figsize=(8,3))
plt.title('Answer prose token count'); plt.xlabel('tokens'); plt.show()
print(df['ans_len'].describe().round(1))
```

## Date Distribution

```python
df['year'] = pd.to_datetime(df['creation_date'],errors='coerce').dt.year
df['year'].value_counts().sort_index().plot(kind='bar',figsize=(10,3),title='Posts by Year')
plt.tight_layout(); plt.show()
```

## Sample QA Pair

```python
row = df.sort_values('question_score',ascending=False).iloc[0]
print('TITLE:', row['question_title'])
print('TAGS:', row['tags_list'])
print('SCORE:', row['question_score'])
print('\nANSWER (first 500 chars):')
print(row['answer_prose'][:500])
```
